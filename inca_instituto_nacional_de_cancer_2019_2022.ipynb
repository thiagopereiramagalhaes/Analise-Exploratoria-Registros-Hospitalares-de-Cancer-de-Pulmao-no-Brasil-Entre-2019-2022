## Entendimento do Negócio

O dataset "INCA_Instituto_Nacional_de_Cancer_1985_2022" constitui um registro histórico e administrativo de morbidade hospitalar, consolidando dados provenientes de Registros Hospitalares de Câncer (RHC) no Brasil. A base descreve a jornada de pacientes oncológicos atendidos em unidades de saúde especializadas, estruturando-se em três dimensões principais: sociodemográfica (incluindo idade, sexo, raça/cor, escolaridade, hábitos de vida e procedência geográfica), clínica (detalhando a localização primária do tumor, tipo histológico, base do diagnóstico e estadiamento da doença) e terapêutica (registrando as datas críticas de diagnóstico e início de tratamento, a modalidade terapêutica adotada e o estado da doença ao final do primeiro tratamento). Essencialmente, o conjunto de dados oferece um panorama sobre a demanda atendida pela rede de oncologia, permitindo análises sobre o perfil epidemiológico dos pacientes e a eficiência do fluxo de assistência à saúde ao longo das últimas décadas.

É importante ressaltar que este projeto concentra sua análise exclusivamente nos CIDs relacionados ao câncer de pulmão e suas metástases, bem como na busca por diagnóstico de câncer. Os códigos CID-O (Classificação Internacional de Doenças para Oncologia) considerados são: C34.0 (Brônquio principal), C34.1 (Lobo superior, brônquio ou pulmão), C34.2 (Lobo médio, brônquio ou pulmão), C34.3 (Lobo inferior, brônquio ou pulmão), C34.8 (Lesão sobreposta do brônquio e pulmão), C34.9 (Brônquio ou pulmão, não especificado), C78.0 (Metástase secundária de pulmão) e Z12.2 (Rastreamento especial para neoplasia de pulmão). Este recorte permite um estudo aprofundado e direcionado sobre a epidemiologia e o perfil de tratamento desses casos específicos no Brasil.

## Entendimento dos Dados

1. **Identificação e Demografia**
* **TPCASO (Tipo de Caso):** Define o vínculo do paciente com a instituição.

  *1: Analítico (Diagnóstico e tratamento feitos na instituição)*

  *2: Não Analítico (Apenas um dos dois ou apenas consulta)*

* **SEXO:** Sexo biológico do paciente.

  *1: Masculino*

  *2: Feminino*

* **IDADE:** Idade do paciente na data da primeira consulta (em anos).

* **RACACOR (Raça/Cor):** Autodeclarada pelo paciente.

  *1: Branca, 2: Preta, 3: Amarela, 4: Parda, 5: Indígena, 9: Sem informação*

* **INSTRUC (Instrução/Escolaridade):** Nível educacional.

  *1: Nenhuma, 2: Fund. Incompleto, 3: Fund. Completo, 4: Médio, 5: Sup. Incompleto, 6: Sup. Completo*

* **ESTCONJ (Estado Conjugal):** Situação marital atual.

  *1: Solteiro, 2: Casado, 3: Viúvo, 4: Separado, 5: União Consensual*

* **OCUPACAO:** Ocupação principal do paciente (Codificação CBO).

* **LOCALNAS:** Sigla da Unidade da Federação (UF) de nascimento.

2. **Dados Clínicos e Características do Tumor**
* **LOCTUPRI (Localização Primária Detalhada):** Código CID-O (4 dígitos) indicando o órgão de origem exato do tumor (ex: C50.9).

* **LOCTUDET (Localização Primária - Categoria):** Código CID-O agrupado (3 dígitos) (ex: C50).

* **TIPOHIST (Tipo Histológico):** Código da morfologia do tumor segundo a CID-O (ex: Carcinoma, Sarcoma).

* **LATERALI (Lateralidade):** Lado do corpo afetado.
  
  *1: Direita, 2: Esquerda, 3: Bilateral, 8: N/A*

* **MAISUMTU (Mais de um tumor):** Indica ocorrência de múltiplos tumores primários.

  *1: Não, 2: Sim, 3: Duvidoso*

* **BASMAIMP (Base Mais Importante do Diagnóstico):** Método principal de confirmação.

  *1: Clínica, 3: Imagem, 5: Citologia, 7: Histologia do tumor primário (Biópsia)*

* **BASDIAGSP (Base de Diagnóstico Simplificada):** Classificação resumida do método diagnóstico.

  *1: Clínico, 2: Não microscópico, 3: Microscópico*

* **EXDIAG:** Exames relevantes realizados para o diagnóstico e planejamento.

* **DIAGANT:** Histórico de diagnóstico e tratamentos anteriores à chegada no hospital.

* **LOCTUPRO:** Localização provável (utilizado quando a localização primária é desconhecida).

3. **Estadiamento e Gravidade da Doença**
* **ESTADIAM (Estadiamento Clínico - Grupo):** Agrupamento da gravidade segundo sistema TNM (Estádio I, II, III, IV).

* **ESTADIAG:** Variação do grupo de estadiamento (frequentemente nulo na base).

* **TNM:** Classificação completa do tumor (T=Tumor, N=Linfonodos, M=Metástase) no momento do diagnóstico.

* **PTNM (Estadiamento Patológico):** Classificação TNM baseada na análise pós-cirúrgica (identificado na análise do CSV).

* **OUTROESTA:** Outros sistemas de estadiamento para tumores que não utilizam TNM (ex: Leucemias, Linfomas).

4. **Tratamento e Evolução**
* **PRITRATH (Primeiro Tratamento Hospitalar):** Modalidade terapêutica inicial.

  *1: Nenhum, 2: Cirurgia, 3: Radioterapia, 4: Quimioterapia, 5: Hormonioterapia*

* **RZNTR (Razão de Não Tratamento):** Motivo caso o tratamento não tenha ocorrido.

  *1: Recusa, 2: Tratou fora, 3: Doença avançada, 6: Óbito*

* **ESTDFIMT (Estado da Doença ao Final do Tratamento):** Resultado do primeiro curso terapêutico.

  *1: Remissão completa, 2: Remissão parcial, 3: Estável, 4: Progressão, 6: Óbito*

* **DATAOBITO:** Data do falecimento do paciente (dd/mm/aaaa).

5. **Localização e Origem**
**ESTADRES:** Sigla da UF de residência habitual do paciente.

* PROCEDEN: Código IBGE do município de residência do paciente.

* CNES: Código do estabelecimento de saúde (hospital) onde o registro foi feito.

* UFUH: UF onde o hospital está localizado.

* MUUH: Município onde o hospital está localizado.

* ORIENC (Origem do Encaminhamento): Como o paciente chegou à unidade.

  *1: SUS, 2: Não SUS (Convênio/Particular), 3: Conta própria*

6. **Hábitos e Histórico**
* **TABAGISM (Tabagismo):** Histórico de consumo de tabaco.

  *1: Nunca, 2: Ex-consumidor, 3: Sim (Fumante atual)*

* **ALCOOLIS (Alcoolismo):** Histórico de consumo de álcool.

  *1: Nunca, 2: Ex-consumidor, 3: Sim (Consumidor atual)*

* **HISTFAMC (Histórico Familiar):** Ocorrência de câncer na família.

  *1: Sim, 2: Não*

7. **Variáveis Temporais**
* **ANOPRIDI:** Ano do primeiro diagnóstico (aaaa).

* **DTDIAGNO:** Data completa do diagnóstico (dd/mm/aaaa).

* **DTINITRT:** Ano de início do tratamento.

* **DATAINITRT:** Data completa do início do tratamento.

* **DTPRICON:** Ano da primeira consulta na instituição.

* **DATAPRICON:** Data completa da primeira consulta.

* **ANTRI:** Ano da triagem.

* **DTTRIAGE:** Data completa da triagem.

* **CLIATEN:** Código da clínica do primeiro atendimento (tabela interna SisRHC).

* **CLITRAT:** Código da clínica de início do tratamento.

* **VALOR_TOT:** Valor total ou custo associado ao procedimento (variável administrativa identificada no CSV).

### Probabilidade Simples

Esta análise calcula a probabilidade geral de dois desfechos importantes em pacientes com câncer de pulmão: a remissão completa da doença e a progressão da doença.

Ao calcular a probabilidade de remissão completa, obtemos uma medida direta da eficácia geral dos tratamentos ou da resiliência da população. Por outro lado, a probabilidade de progressão da doença nos dá uma indicação da gravidade e da dificuldade em controlar a doença na população estudada.

```python
print("Análise de Probabilidade Simples:\n")

# Filtrar dados onde o estado final da doença não é 'Não Informado'
filtered_df_simple = df[df['ds_estado_doenca_final_tratamento'] != 'Não Informado'].copy()

total_patients_with_outcome = filtered_df_simple.shape[0]

if total_patients_with_outcome == 0:
    print("Não há dados suficientes para realizar a análise após filtrar 'Não Informado' na coluna de estado da doença.")
else:
    # Calcular a probabilidade de Remissão Completa
    remissao_completa_count = filtered_df_simple[filtered_df_simple['ds_estado_doenca_final_tratamento'] == 'Remissão completa'].shape[0]
    prob_remissao_completa = remissao_completa_count / total_patients_with_outcome

    # Calcular a probabilidade de Progressão da Doença
    progressao_count = filtered_df_simple[filtered_df_simple['ds_estado_doenca_final_tratamento'] == 'Progressão'].shape[0]
    prob_progressao = progressao_count / total_patients_with_outcome

    print(f"Probabilidade de Remissão Completa: {prob_remissao_completa:.4f} ({prob_remissao_completa:.2%})")
    print(f"Probabilidade de Progressão da Doença: {prob_progressao:.4f} ({prob_progressao:.2%})")

    print("\nInterpretação:")
    print(f"A probabilidade geral de um paciente alcançar a remissão completa é de {prob_remissao_completa:.2%}.")
    print(f"A probabilidade geral de um paciente apresentar progressão da doença é de {prob_progressao:.2%}.")
```

Esta análise calcula a probabilidade geral de um paciente ir a óbito na população total do dataset, considerando apenas aqueles que têm um desfecho informado.

 A probabilidade de óbito é um dos indicadores mais críticos na análise de saúde pública e oncologia. Ela fornece uma visão da letalidade da doença na população geral estudada. Conhecer essa probabilidade nos permite entender a magnitude do problema e serve como um ponto de referência para análises mais detalhadas, como a probabilidade de óbito em diferentes faixas etárias ou com diferentes tipos de tratamento.

```python
print("Análise de Probabilidade Simples: Óbito na População Total\n")

# Filtrar dados onde o estado final da doença não é 'Não Informado'
filtered_df_total = df[df['ds_estado_doenca_final_tratamento'] != 'Não Informado'].copy()

total_patients_with_outcome = filtered_df_total.shape[0]

if total_patients_with_outcome == 0:
    print("Não há dados suficientes para realizar a análise após filtrar 'Não Informado' na coluna de estado da doença.")
else:
    # Calcular o número de óbitos na população total com desfecho informado
    obito_total_count = filtered_df_total[filtered_df_total['ds_estado_doenca_final_tratamento'] == 'Óbito'].shape[0]
    prob_obito_total = obito_total_count / total_patients_with_outcome

    print(f"Total de pacientes com desfecho informado: {total_patients_with_outcome}")
    print(f"Número de óbitos na população total: {obito_total_count}")
    print(f"Probabilidade de óbito na população total: {prob_obito_total:.4f} ({prob_obito_total:.2%})")

    print("\nInterpretação:")
    print(f"A probabilidade geral de um paciente ir a óbito, caso desenvolva câncer (e tenha um desfecho informado), é de {prob_obito_total:.2%}.")
```

### Probabilidade Condicional

 A probabilidade condicional permite investigar como a chance de um evento (óbito) muda dada a ocorrência de outro evento (faixa etária, origem de encaminhamento). Isso é crucial para identificar grupos de risco ou sistemas de acesso que podem estar associados a piores prognósticos, informando políticas de saúde e intervenções específicas.

### Teorema de Bayes

Neste caso, permite quantificar a relação entre um fator de risco (tabagismo) e um desfecho (progressão da doença), fornecendo um insight mais aprofundado do que uma simples correlação. A análise aqui revelou um resultado contraintuitivo, destacando a importância de uma interpretação cuidadosa e a possível necessidade de considerar mais variáveis.

```python
print("Análise utilizando o Teorema de Bayes:\n")

# Definição dos Eventos:
# Evento A: O paciente apresenta 'Progressão' no estado final da doença (ds_estado_doenca_final_tratamento).
# Evento B: O paciente é 'Sim (Fumante atual)' no status de tabagismo (ds_tabagismo).

# Filtrar dados relevantes, removendo 'Não Informado' que não representa uma categoria válida para os eventos.
filtered_df_bayes = df[
    (df['ds_estado_doenca_final_tratamento'] != 'Não Informado') &
    (df['ds_tabagismo'] != 'Não Informado')
].copy()

total_relevant_patients_bayes = filtered_df_bayes.shape[0]

if total_relevant_patients_bayes == 0:
    print("Não há dados suficientes para realizar a análise de Bayes após filtrar 'Não Informado' para as variáveis selecionadas.")
else:
    # 1. Calcular P(A): Probabilidade a priori de Progressão da Doença
    event_A_count = filtered_df_bayes[filtered_df_bayes['ds_estado_doenca_final_tratamento'] == 'Progressão'].shape[0]
    P_A = event_A_count / total_relevant_patients_bayes

    # 2. Calcular P(B): Probabilidade a priori de ser Fumante Atual
    event_B_count = filtered_df_bayes[filtered_df_bayes['ds_tabagismo'] == 'Sim (Fumante atual)'].shape[0]
    P_B = event_B_count / total_relevant_patients_bayes

    # 3. Calcular P(A e B): Probabilidade de Progressão DA DOENÇA E ser Fumante Atual
    event_A_and_B_count = filtered_df_bayes[
        (filtered_df_bayes['ds_estado_doenca_final_tratamento'] == 'Progressão') &
        (filtered_df_bayes['ds_tabagismo'] == 'Sim (Fumante atual)')
    ].shape[0]
    P_A_and_B = event_A_and_B_count / total_relevant_patients_bayes

    # 4. Calcular P(B|A): Probabilidade de ser Fumante Atual dado que há Progressão da Doença
    # P(B|A) = P(A e B) / P(A)
    P_B_given_A = P_A_and_B / P_A if P_A > 0 else 0

    # 5. Aplicar o Teorema de Bayes para calcular P(A|B):
    # P(A|B) = P(B|A) * P(A) / P(B)
    # Probabilidade de Progressão da Doença dado que é Fumante Atual
    P_A_given_B = (P_B_given_A * P_A) / P_B if P_B > 0 else 0

    print(f"Evento A: Paciente apresenta 'Progressão' no estado final da doença.")
    print(f"Evento B: Paciente é 'Sim (Fumante atual)' no status de tabagismo.")
    print("-" * 70)
    print(f"P(A) = Probabilidade de Progressão da Doença: {P_A:.4f} ({P_A:.2%})")
    print(f"P(B) = Probabilidade de ser Fumante Atual: {P_B:.4f} ({P_B:.2%})")
    print(f"P(A e B) = Probabilidade de Progressão da Doença E ser Fumante Atual: {P_A_and_B:.4f} ({P_A_and_B:.2%})")
    print(f"P(B|A) = Probabilidade de ser Fumante Atual | Progressão da Doença: {P_B_given_A:.4f} ({P_B_given_A:.2%})")
    print("-" * 70)
    print(f"P(A|B) = Probabilidade de Progressão da Doença | Fumante Atual (via Teorema de Bayes): {P_A_given_B:.4f} ({P_A_given_B:.2%})")

    print("\nInterpretação:")
    print(f"A probabilidade geral de um paciente apresentar progressão da doença é de {P_A:.2%}.")
    print(f"A probabilidade de um paciente ser fumante atual é de {P_B:.2%}.")
    print(f"A probabilidade de um paciente apresentar progressão da doença, DADO que ele é um fumante atual, é de {P_A_given_B:.2%}.")

    if P_A > 0:
        if P_A_given_B > P_A:
            print("Isso sugere que o tabagismo atual está associado a uma *maior* probabilidade de progressão da doença em comparação com a probabilidade geral de progressão.")
        elif P_A_given_B < P_A:
            print("Isso sugere que o tabagismo atual está associado a uma *menor* probabilidade de progressão da doença em comparação com a probabilidade geral de progressão.")
        else:
            print("Isso sugere que o tabagismo atual não altera significativamente a probabilidade de progressão da doença.")
```

### Normalização dos Dados

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

print("Análise de Normalização de Dados para Machine Learning (Apenas Idade)\n")

# 1. Escolha de variáveis numéricas para normalização
selected_columns = ['nr_idade']
print(f"Variáveis selecionadas para normalização: {selected_columns}")

# 2. Tratamento de valores ausentes e preparação dos dados para escala
# Criar uma cópia dos dados selecionados para evitar modificar o DataFrame original diretamente
# e preencher NaNs.
data_for_scaling = df[selected_columns].copy()

for col in selected_columns:
    if data_for_scaling[col].isnull().any():
        median_val = data_for_scaling[col].median()
        data_for_scaling[col] = data_for_scaling[col].fillna(median_val)
        print(f"Preenchendo NaNs da coluna '{col}' com a mediana ({median_val}) para a etapa de normalização.")

# 3. Aplicar StandardScaler
scaler_standard = StandardScaler()
data_standard_scaled = scaler_standard.fit_transform(data_for_scaling)
df_standard_scaled = pd.DataFrame(data_standard_scaled, columns=[f'{col}_standard' for col in selected_columns], index=data_for_scaling.index)
print("\nStandardScaler aplicado.")

# 4. Aplicar MinMaxScaler
scaler_minmax = MinMaxScaler()
data_minmax_scaled = scaler_minmax.fit_transform(data_for_scaling)
df_minmax_scaled = pd.DataFrame(data_minmax_scaled, columns=[f'{col}_minmax' for col in selected_columns], index=data_for_scaling.index)
print("MinMaxScaler aplicado.")

# 5. Visualizar e comparar as distribuições em uma única figura
fig, axes = plt.subplots(len(selected_columns), 3, figsize=(18, 5 * len(selected_columns)))
fig.suptitle('Comparação das Distribuições de Variáveis Numéricas (Apenas Idade)', fontsize=16)

# Ensure axes is always 2D for consistent indexing, especially when len(selected_columns) is 1
if len(selected_columns) == 1:
    axes = np.expand_dims(axes, axis=0)

for i, col in enumerate(selected_columns):
    # Original Distribution
    sns.histplot(data_for_scaling[col], kde=True, color='blue', ax=axes[i, 0])
    axes[i, 0].set_title(f'Original: {col}')
    axes[i, 0].set_xlabel('Valor Original')
    axes[i, 0].set_ylabel('Frequência')

    # StandardScaler Distribution
    sns.histplot(df_standard_scaled[f'{col}_standard'], kde=True, color='green', ax=axes[i, 1])
    axes[i, 1].set_title(f'Standard Scaled: {col}')
    axes[i, 1].set_xlabel('Valor (Standard Scaled)')
    axes[i, 1].set_ylabel('Frequência')

    # MinMaxScaler Distribution
    sns.histplot(df_minmax_scaled[f'{col}_minmax'], kde=True, color='red', ax=axes[i, 2])
    axes[i, 2].set_title(f'MinMax Scaled: {col}')
    axes[i, 2].set_xlabel('Valor (MinMax Scaled)')
    axes[i, 2].set_ylabel('Frequência')

plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Ajusta o layout para evitar sobreposição do título
plt.show()

print("\nInterpretação:\n")
print("Os gráficos acima ilustram o efeito do StandardScaler e do MinMaxScaler na distribuição da variável selecionada. Para a idade, são mostrados:")
print("-   A distribuição original: Reflete a escala e forma iniciais dos dados.")
print("    - StandardScaler: Transforma os dados para ter média 0 e desvio padrão 1. A forma da distribuição é preservada, mas é centralizada e redimensionada.")
print("    - MinMaxScaler: Redimensiona os dados para um intervalo fixo (geralmente [0, 1]). A forma da distribuição também é preservada, mas os valores são mapeados para o novo intervalo.")
print("A escolha entre StandardScaler e MinMaxScaler depende do algoritmo de Machine Learning e das características dos seus dados. StandardScaler é geralmente preferido quando a distribuição original é aproximadamente normal ou quando os algoritmos se beneficiam de dados com média zero e variância unitária (ex: PCA, regressão linear). MinMaxScaler é útil quando a normalização para um intervalo específico é desejada (ex: redes neurais) ou quando os algoritmos são sensíveis a outliers e o intervalo máximo/mínimo é importante para a performance (embora outliers ainda afetem o escalamento).")
```

### Distribuições Amostrais e Teorema do Limite Central

Esta análise envolve a simulação de múltiplas amostragens aleatórias da população de pacientes para demonstrar que a distribuição das médias amostrais tende a uma distribuição normal (curva em sino), independentemente da distribuição original dos dados (que pode ser assimétrica).

```python
print("Análise de Distribuições Amostrais e Teorema do Limite Central (Óbitos)\n")

# Filtrar dados para incluir apenas óbitos
deaths_df = df[df['ds_estado_doenca_final_tratamento'] == 'Óbito'].copy()

# Remover valores nulos da coluna de idade dos óbitos para o cálculo
idade_obito_clean = deaths_df['nr_idade'].dropna()

# Verificar se há dados suficientes após a filtragem
if idade_obito_clean.empty:
    print("Não há dados suficientes de idade para pacientes que foram a óbito para realizar esta análise.")
else:
    # Parâmetros da simulação
    num_samples = 1000  # Número de amostras a serem tiradas
    sample_size = 30    # Tamanho de cada amostra

    # Garantir que o tamanho da amostra não excede o número de dados disponíveis
    if sample_size > len(idade_obito_clean):
        sample_size = len(idade_obito_clean) # Ajusta o tamanho da amostra se for maior que os dados disponíveis
        print(f"Ajustado o tamanho da amostra para {sample_size} devido ao número limitado de dados de óbitos.")

    # Lista para armazenar as médias das amostras
    media_das_amostras_obito = []

    # Realizar a amostragem e calcular as médias
    for _ in range(num_samples):
        sample = idade_obito_clean.sample(sample_size, random_state=np.random.randint(0, 1000000))
        media_das_amostras_obito.append(sample.mean())

    # Converter a lista de médias para um array numpy para facilitar a análise
    media_das_amostras_obito = np.array(media_das_amostras_obito)

    # Plotar a distribuição original da idade dos óbitos
    plt.figure(figsize=(15, 6))
    plt.subplot(1, 2, 1)
    sns.histplot(idade_obito_clean, bins=30, kde=True, color='skyblue', edgecolor='black')
    plt.title('Distribuição Original da Idade dos Pacientes que Foram a Óbito')
    plt.xlabel('Idade')
    plt.ylabel('Frequência')

    # Plotar a distribuição das médias das amostras de óbitos
    plt.subplot(1, 2, 2)
    sns.histplot(media_das_amostras_obito, bins=30, kde=True, color='lightcoral', edgecolor='black')
    plt.title('Distribuição das Médias Amostrais (Teorema do Limite Central) - Óbitos')
    plt.xlabel('Média da Idade da Amostra (Óbitos)')
    plt.ylabel('Frequência')

    plt.tight_layout()
    plt.show()

    print("\nInterpretação:\n")
    print(f"Média da população original (idade dos óbitos): {idade_obito_clean.mean():.2f}")
    print(f"Média das médias amostrais (idade dos óbitos): {media_das_amostras_obito.mean():.2f}")
    print("\nObservamos que a distribuição das médias amostrais para a idade dos pacientes que foram a óbito também se aproxima de uma distribuição normal, com a média das médias amostrais sendo muito próxima da média real da idade dos óbitos. Isso demonstra a validade do Teorema do Limite Central mesmo para este subgrupo específico da população.")
```

### Intervalo de Confiança

Esta etapa envolve o cálculo de intervalos (tipicamente de 90% ou 95%) para estimar onde se encontra o verdadeiro parâmetro da população (como a média de idade ou o tempo médio de tratamento) com base na amostra do dataset.

O Intervalo de Confiança fornece a precisão da estimativa. Dizer que a "idade média é 65 anos" é menos informativo do que dizer "temos 95% de confiança de que a idade média da população está entre 65,43 e 65,75 anos". Isso é crucial para o planejamento de recursos, pois permite prever cenários de demanda

```python
print("Análise de Intervalo de Confiança: Idade Média\n")

# Remover valores nulos da coluna de idade para o cálculo
idade_clean = df['nr_idade'].dropna()

# 1. Calcular a média e o desvio padrão da amostra
media_idade = idade_clean.mean()
desvio_padrao_idade = idade_clean.std()
tamanho_amostra = idade_clean.count()

# 2. Definir o nível de confiança (ex: 95%)
nivel_confianca = 0.95

# 3. Calcular o erro padrão da média
erro_padrao = desvio_padrao_idade / np.sqrt(tamanho_amostra)

# 4. Calcular o valor crítico (z-score para grandes amostras)
# Para 95% de confiança, alpha = 0.05, alpha/2 = 0.025. O z-score é para 1 - alpha/2 = 0.975
valor_critico_z = stats.norm.ppf(1 - (1 - nivel_confianca) / 2)

# 5. Calcular a margem de erro
margem_erro = valor_critico_z * erro_padrao

# 6. Calcular o Intervalo de Confiança
intervalo_inferior = media_idade - margem_erro
intervalo_superior = media_idade + margem_erro

print(f"Média da idade da amostra: {media_idade:.2f} anos")
print(f"Desvio padrão da idade da amostra: {desvio_padrao_idade:.2f}")
print(f"Tamanho da amostra: {tamanho_amostra}")
print(f"Nível de Confiança: {nivel_confianca*100:.0f}%")
print(f"Margem de Erro: {margem_erro:.2f}")
print(f"Intervalo de Confiança ({nivel_confianca*100:.0f}%): [{intervalo_inferior:.2f}, {intervalo_superior:.2f}] anos")

print("\nInterpretação:")
print(f"Podemos afirmar, com {nivel_confianca*100:.0f}% de confiança, que a idade média real da população de pacientes com câncer no Brasil, a partir da qual esta amostra foi retirada, está entre {intervalo_inferior:.2f} e {intervalo_superior:.2f} anos.")
```
